Part B

```python
dataset={
    'population':population,
    'GDP':gdp,
    'Internet':internet_users,
    'Literacy':literacy_rate
}
for name,df in dataset.items():
  print('\nDataset:', name)
  print(df.shape)
  print(df.columns)
  print('\nData Types:')
  print(df.dtypes)
  print('\nMissing Values:')
  print(df.isnull().sum())
  print(df.duplicated().sum())
```

Part C

```python
gdp.rename(columns={'Country':'country'},inplace=True)
literacy_rate.rename(columns={'name':'country'},inplace=True)
```

task 7


```python
population["population"] = pd.to_numeric(population["population"])

gdp["GDP"] = pd.to_numeric(gdp["GDP"])

internet_users["internet_users"] = pd.to_numeric(internet_users["internet_users"])

literacy_rate["literacy_rate"] = pd.to_numeric(literacy_rate["literacy_rate"])
```

```python
print(population.dtypes)
print(gdp.dtypes)
print(internet_users.dtypes)
print(literacy_rate.dtypes)
```

Task 8

```python
merged_df = pd.merge(population, gdp, on="country", how="inner")

merged_df = pd.merge(merged_df, internet_users, on="country", how="inner")

merged_df = pd.merge(merged_df, literacy_rate, on="country", how="inner")
```

```python
merged_df.head()
```

```python
merged_df.shape
```

task 9

```python
merged_df["internet_penetration"] = (
    merged_df["internet_users"] / merged_df["population"]) * 100

merged_df["internet_penetration"] = merged_df["internet_penetration"].round(2)
```

```python
merged_df["GDP_per_capita"] = (
    merged_df["GDP"] / merged_df["population"])

merged_df["GDP_per_capita"] = merged_df["GDP_per_capita"].round(2)
```

```python
merged_df.head()
```

task 10

```python
conn = sqlite3.connect("country_statistics.db")
```

```python
merged_df.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)
```

task 11

```python
result = pd.read_sql_query(
    "SELECT * FROM country_statistics LIMIT 5",
    conn
)

result
```
